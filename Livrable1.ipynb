# Plan de livrable

## 1. Représentation formelle des données du problème
- Décrire les variables, les paramètres et les entités du problème.
- Formaliser les données sous forme de vecteurs, matrices, graphes, ensembles, etc.
- Préciser les types, les dimensions et les éventuelles relations entre les données.

## 2. Contraintes et intégration
- Énumérer les contraintes du problème (capacités, ressources, temps, budget, etc.).
- Formaliser chaque contrainte avec des équations ou des inégalités.
- Expliquer comment ces contraintes sont intégrées dans le modèle.

## 3. Représentation formelle de l’objectif à optimiser
- Décrire la fonction objectif à minimiser ou maximiser.
- Préciser le sens de l’optimisation et les variables de décision.
- Donner la formulation mathématique complète de l’objectif.

## 4. Complexité théorique
- Analyser la complexité du problème à partir de la formulation formelle.
- Identifier si le problème est NP-complet, NP-difficile, ou appartient à une autre classe.
- Expliquer les implications en termes de temps de calcul et d’algorithmes approchés ou exacts.

## 5. Références bibliographiques
- Lister les sources, articles, livres ou documents utilisés.
- Inclure des références formatées (auteur, année, titre, publication).
- Ajouter des liens vers les ressources pertinentes si nécessaire.
