## 2. Représentation formelle de l’objectif à optimiser R
**Fonction objectif :**  

$\min_{Z} \sum_{i \in V} \sum_{j \in V} \sum_{k \in K} c_{ij} \times x_{ijk}$
 
Dans ce problème, on cherche à définir une planification de routes pour plusieurs véhicules de façon à visiter chaque sommet ($i \in V$) une seule fois, tout en minimisant le coût total associé aux déplacements sur les arêtes ( $ij \in E$ ).
 
### Sens de l’optimisation
Le terme **"min"** indique que l’on souhaite **minimiser le coût total du parcours**, autrement dit trouver la configuration de routes qui entraîne le **coût global le plus faible**.  
Chaque véhicule ($k \in K$) suit un itinéraire reliant des sommets ($i, j \in V$), et le coût de déplacement entre ces sommets est donné par ($c_{ij}$). La fonction objectif additionne tous ces coûts pour l’ensemble des véhicules.
 
### Variables de décision
*répétition avec la partie représentation formelle des données du problème*
La variable de décision principale est :
$$
x_{ijk} = 
\begin{cases} 
1 & \text{si le véhicule } k \text{ parcourt l’arête } (i,j),\\ 
0 & \text{sinon.}
\end{cases}
$$
Elle permet de déterminer quels véhicules passent par quelles arêtes, donc de définir l’ensemble des trajets à effectuer.
